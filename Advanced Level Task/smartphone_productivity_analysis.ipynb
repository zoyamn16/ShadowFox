```python
# Smartphone Usage & Productivity Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("smartphone_usage.csv")

print("Dataset Preview:")
print(df.head())

print("\nDataset Info:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

plt.figure()
sns.scatterplot(x="daily_screen_time", y="productivity_score", data=df)
plt.title("Daily Screen Time vs Productivity Score")
plt.xlabel("Daily Screen Time (hours)")
plt.ylabel("Productivity Score")
plt.show()

plt.figure()
sns.lineplot(x="daily_screen_time", y="sleep_hours", data=df)
plt.title("Daily Screen Time vs Sleep Hours")
plt.xlabel("Daily Screen Time (hours)")
plt.ylabel("Sleep Hours")
plt.show()

plt.figure()
sns.boxplot(x="stress_level", y="productivity_score", data=df)
plt.title("Productivity Score by Stress Level")
plt.xlabel("Stress Level")
plt.ylabel("Productivity Score")
plt.show()

plt.figure()
sns.scatterplot(x="social_media_time", y="productivity_score", data=df)
plt.title("Social Media Time vs Productivity Score")
plt.xlabel("Social Media Time (hours)")
plt.ylabel("Productivity Score")
plt.show()

plt.figure(figsize=(8,6))
sns.heatmap(
    df[["daily_screen_time", "social_media_time", "sleep_hours", "productivity_score"]].corr(),
    annot=True,
    cmap="coolwarm"
)
plt.title("Correlation Between Smartphone Usage and Productivity Factors")
plt.show()

print("\nConclusion:")
print("1. Higher daily screen time is associated with lower productivity scores.")
print("2. Increased smartphone usage reduces sleep hours, negatively impacting productivity.")
print("3. Social media usage shows a negative relationship with productivity.")
print("4. High stress levels correspond to lower productivity scores.")
print("5. Managing screen time, improving sleep, and reducing stress can enhance productivity.")
```
